```python
import pandas as pd

def band(p):
    if p < 50: return 'Low'
    elif p < 75: return 'Medium'
    else: return 'High'

df_raw = pd.read_csv('../data/attendance_dataset-V3_500_rows.csv')
df_raw['band'] = df_raw['Attendance_Percentage'].apply(band)
print(df_raw['band'].value_counts())
print()
print(df_raw['Attendance_Percentage'].describe())
```
